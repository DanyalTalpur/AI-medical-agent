# 🩺 Medical Report AI Agent

Welcome to the **Medical Report AI Agent** notebook. This application combines document extraction, a rule-based NLP parser, interactive visualization, and Google Gemini LLM capabilities to analyze medical lab reports and provide clear, patient-friendly explanations.

### Key Features
1. **Multi-format Support:** Analyzes digital PDFs, scanned PDFs, and image uploads (PNG, JPG, JPEG).
2. **Hybrid Parser:**
   - **Offline Mode:** Uses an advanced regex parser and standard medical reference dictionary to extract 20+ common lab tests (Glucose, Hb, WBC, RBC, Cholesterol, Kidney/Liver markers) and identify out-of-range "Red Flags".
   - **Online Mode:** Connects to the **Google Gemini API** (using `gemini-2.5-flash`) to perform high-fidelity medical summarization, clinical translation, and multimodal OCR on scanned files.
3. **Visual Range Tracker:** Generates an interactive Plotly chart showing where your values lie within the normal reference intervals.
4. **Interactive Chatbot:** A conversational agent that acts as a patient assistant, allowing you to ask follow-up questions about your report in real time.
5. **Modern Dark/Light UI:** An elegant, responsive web interface built using Gradio.

```python
def extract_metadata(text):
    """Parses basic patient metadata from report text."""
    metadata = {"Name": "Unknown", "Age": "Unknown", "Gender": "Unknown", "Date": "Unknown"}
    lines = text.split("\n")
    for line in lines:
        name_match = re.search(r"\b(?:patient\s+)?name\b\s*[:\-]?\s*([a-zA-Z\s\.\,\-]+)", line, re.IGNORECASE)
        if name_match:
            metadata["Name"] = name_match.group(1).strip()
            
        age_match = re.search(r"\bage\b\s*[:\-]?\s*(\d+)", line, re.IGNORECASE)
        if age_match:
            metadata["Age"] = age_match.group(1).strip()
            
        gender_match = re.search(r"\bgender\b\s*[:\-]?\s*(\bmale\b|\bfemale\b|\bm\b|\bf\b)", line, re.IGNORECASE)
        if gender_match:
            gen = gender_match.group(1).strip().lower()
            metadata["Gender"] = "Male" if gen in ["male", "m"] else "Female"
            
        date_match = re.search(r"\b(?:date|collected|reported)\b\s*[:\-]?\s*([\d\/\-\s\w,]+)", line, re.IGNORECASE)
        if date_match:
            metadata["Date"] = date_match.group(1).strip()
            
    return metadata

def extract_text_from_pdf(pdf_path):
    """Extracts raw text from digital PDF document pages."""
    text = ""
    try:
        doc = fitz.open(pdf_path)
        for page in doc:
            text += page.get_text() + "\n"
        doc.close()
    except Exception as e:
        print(f"Error reading PDF: {e}")
    return text

def parse_lab_values_from_text(text):
    """Scans raw report text for common medical lab tests using regex and standard dictionary."""
    results = {}
    lines = text.split("\n")
    
    for name, info in LAB_DICTIONARY.items():
        for pattern in info["synonyms"]:
            rx = re.compile(pattern, re.IGNORECASE)
            for line in lines:
                match = rx.search(line)
                if match:
                    # Slice line post-match to search for value
                    post_match_text = line[match.end():]
                    # Find numbers (integers or floats)
                    post_numbers = re.findall(r"\b\d+(?:\.\d+)?\b", post_match_text)
                    if post_numbers:
                        # The first number after the test name is the observed value
                        try:
                            val = float(post_numbers[0])
                        except ValueError:
                            continue
                        
                        # Try to find a custom reference range (e.g. 70-99 or 70 - 100)
                        ref_low = info["low"]
                        ref_high = info["high"]
                        custom_range_match = re.search(r"(\d+(?:\.\d+)?)\s*[-–—]\s*(\d+(?:\.\d+)?)", post_match_text)
                        
                        if custom_range_match:
                            try:
                                ref_low = float(custom_range_match.group(1))
                                ref_high = float(custom_range_match.group(2))
                            except ValueError:
                                pass
                        else:
                            # Check for operator bounds like '< 100' or '> 40'
                            limit_match = re.search(r"([<>])\s*(\d+(?:\.\d+)?)", post_match_text)
                            if limit_match:
                                try:
                                    op = limit_match.group(1)
                                    lim_val = float(limit_match.group(2))
                                    if op == "<":
                                        ref_low = 0.0
                                        ref_high = lim_val
                                    else:
                                        ref_low = lim_val
                                        ref_high = 999999.0
                                except ValueError:
                                    pass
                        
                        # Determine clinical status
                        status = "Normal"
                        if val < ref_low:
                            status = "Low"
                        elif val > ref_high:
                            status = "High"
                            
                        results[name] = {
                            "value": val,
                            "low": ref_low,
                            "high": ref_high,
                            "unit": info["unit"],
                            "status": status,
                            "description": info["description"]
                        }
                        break  # Successfully found and parsed this test; break line loop
    return results

def generate_offline_summary(parsed_results, metadata):
    """Generates a detailed, patient-friendly markdown explanation when no API key is provided."""
    if not parsed_results:
        return """
# 🩺 Lab Report Analysis (Offline Mode)
**No recognizable lab markers were detected in the text.**

*Note: The offline parser scans for specific names (e.g., Fasting Glucose, Hemoglobin, WBC, Cholesterol, Creatinine). Please check the text in the editor, edit values, or load the Sample Report.*
"""
    
    summary = f"""# 🩺 Lab Report Analysis (Offline Mode)
**Patient:** {metadata.get('Name', 'Unknown')} | **Age:** {metadata.get('Age', 'Unknown')} | **Gender:** {metadata.get('Gender', 'Unknown')}
**Date of Report:** {metadata.get('Date', 'Unknown')}

## 📊 Summary of Findings
The system identified **{len(parsed_results)}** clinical tests.
- **Normal Values:** {sum(1 for v in parsed_results.values() if v['status'] == 'Normal')}
- **Out-of-Range Flags:** {sum(1 for v in parsed_results.values() if v['status'] != 'Normal')}

---

## 🔍 Detailed Lab Review
"""
    
    red_flags = []
    normal_items = []
    
    for name, data in parsed_results.items():
        if data['status'] != 'Normal':
            red_flags.append(name)
            summary += f"\n### ⚠️ {name} is {data['status'].upper()}\n"
            summary += f"- **Observed Value:** `{data['value']} {data['unit']}` (Standard: `{data['low']} - {data['high']} {data['unit']}`)\n"
            summary += f"- **Meaning:** {data['description']}\n"
        else:
            normal_items.append(f"- **{name}**: `{data['value']} {data['unit']}` (Ref: {data['low']} - {data['high']} {data['unit']}) — *Normal*")
            
    if red_flags:
        summary += "\n## 🚨 Clinical Red Flags\nThe following values were identified outside normal limits. We recommend bringing these up in your next appointment:\n"
        for rf in red_flags:
            summary += f"- **{rf}** ({parsed_results[rf]['status']})\n"
            
    if normal_items:
        summary += "\n## ✅ Normal Parameters\n"
        for item in normal_items:
            summary += item + "\n"
            
    summary += """
---
## 💡 Next Steps & Suggested Questions for Your Doctor
- Bring a copy of this report to your medical provider.
- Ask:
  1. *'What could be causing the abnormal levels of my markers?'*
  2. *'Do you recommend repeating these tests in 3 to 6 months?'*
  3. *'Are there any immediate lifestyle or nutritional adjustments I should make?'*
"""
    return summary

def generate_red_flags_markdown(parsed_results):
    """Produces the HTML/markdown text warning block for abnormal values."""
    abnormal = {k: v for k, v in parsed_results.items() if v['status'] != 'Normal'}
    if not abnormal:
        return "### ✅ All analyzed lab parameters are within normal reference ranges!"
        
    markdown = "### 🚨 Abnormal Values (Red Flags):\n"
    for name, data in abnormal.items():
        markdown += f"- **{name}**: **{data['value']} {data['unit']}** (Ref: {data['low']} - {data['high']}) — *{data['status']}*\n"
    return markdown
```

```python
def convert_pdf_to_images(pdf_path):
    """Converts a PDF file's pages into PNG image byte list for Gemini multimodal input."""
    images_data = []
    try:
        doc = fitz.open(pdf_path)
        for page in doc:
            pix = page.get_pixmap(dpi=150)
            img_data = pix.tobytes("png")
            images_data.append(img_data)
        doc.close()
    except Exception as e:
        print(f"Error converting PDF to image: {e}")
    return images_data

def analyze_report_with_gemini(api_key, raw_text, parsed_data):
    """Calls Gemini to explain a digital report's text and structured parameters."""
    if not api_key:
        return "Please enter a Gemini API Key to enable advanced AI analysis."
    try:
        genai.configure(api_key=api_key)
        model = genai.GenerativeModel('gemini-2.5-flash')
        
        prompt = f"""
        You are a highly competent, empathetic Medical AI Agent.
        Analyze this patient's lab report.
        
        Raw Text of Report:
        ---
        {raw_text}
        ---
        
        Structured Extracted Lab Values:
        ---
        {parsed_data}
        ---
        
        Please generate a report with:
        1. An empathetic Executive Summary of the patient's lab report.
        2. A clear breakdown of "Red Flags" (High/Low values). For each:
           - Explain the marker in patient-friendly terms (analogy-based if helpful).
           - Offer potential non-critical and lifestyle variables that could cause variations (stress, hydration, diet).
           - Suggest general positive dietary/lifestyle improvements.
        3. Highlight the Normal Parameters and express reassurance.
        4. 3-4 pointed, specific questions for their doctor.
        
        Requirements:
        - Format with clean Markdown (bold text, bullet points, headers).
        - Start with a reassuring Medical Disclaimer.
        """
        response = model.generate_content(prompt)
        return response.text
    except Exception as e:
        return f"Error communicating with Gemini: {str(e)}"

def ocr_with_gemini(api_key, file_path):
    """Sends image or PDF pages to Gemini for plain OCR text extraction."""
    try:
        genai.configure(api_key=api_key)
        model = genai.GenerativeModel('gemini-2.5-flash')
        
        _, ext = os.path.splitext(file_path)
        ext = ext.lower()
        
        contents = []
        if ext == '.pdf':
            images_data = convert_pdf_to_images(file_path)
            if not images_data:
                return "Error: Failed to extract pages from PDF."
            for img_bytes in images_data:
                contents.append({"mime_type": "image/png", "data": img_bytes})
        elif ext in ['.png', '.jpg', '.jpeg']:
            with open(file_path, "rb") as f:
                img_bytes = f.read()
            mime_type = "image/png" if ext == '.png' else "image/jpeg"
            contents.append({"mime_type": mime_type, "data": img_bytes})
        else:
            return f"Error: Unsupported file type: {ext}"
            
        prompt = """
        You are an expert OCR system. Extract all readable text from the attached lab report.
        Do not summarize, do not analyze, do not add comments.
        Just output the extracted text, retaining as much layout and alignment as possible.
        """
        contents.append(prompt)
        response = model.generate_content(contents)
        return response.text
    except Exception as e:
        return f"Error running OCR with Gemini: {str(e)}"

def chatbot_respond(api_key, message, history, report_context):
    """Manages the chatbot conversation, referencing the report content."""
    if not api_key:
        # Fallback offline chat simulation
        history.append((message, "🤖 [Offline Mode] To ask personalized questions about your report and get AI responses, please provide a Gemini API Key in the configuration panel. Without it, I cannot process free-form medical queries."))
        return "", history
        
    try:
        genai.configure(api_key=api_key)
        
        system_instruction = f"""
        You are an empathetic, knowledgeable, and professional Medical AI Agent.
        The user has uploaded a lab report. Below is the parsed summary of their report:
        ---
        {report_context}
        ---
        
        Guidelines:
        1. Provide reassuring, accurate, and easy-to-understand explanations of medical terms, tests, or results.
        2. Never diagnose the patient, recommend prescription dosages, or tell them a treatment is 'guaranteed'.
        3. Advise them to consult their primary physician for health choices.
        4. If they ask things unrelated to health/medicine, politely redirect them back to their report.
        """
        
        model = genai.GenerativeModel('gemini-2.5-flash', system_instruction=system_instruction)
        
        # Format conversation history
        gemini_history = []
        for user_msg, agent_msg in history:
            if "[Offline Mode]" in agent_msg:
                continue
            gemini_history.append({"role": "user", "parts": [user_msg]})
            gemini_history.append({"role": "model", "parts": [agent_msg]})
            
        chat = model.start_chat(history=gemini_history)
        response = chat.send_message(message)
        
        history.append((message, response.text))
        return "", history
    except Exception as e:
        history.append((message, f"Error: {str(e)}"))
        return "", history
```

```python
def generate_plotly_charts(parsed_results):
    """Creates a custom, highly intuitive bar visualization showing values relative to range boundaries."""
    if not parsed_results:
        return None
        
    params = list(parsed_results.keys())
    
    # Set up Subplots
    fig = make_subplots(
        rows=len(params), cols=1,
        shared_xaxes=False,
        vertical_spacing=max(0.02, 0.5 / len(params)),
        subplot_titles=[f"<b>{p}</b> ({parsed_results[p]['unit']})" for p in params]
    )
    
    for i, param in enumerate(params):
        data = parsed_results[param]
        val = data["value"]
        low = data["low"]
        high = data["high"]
        unit = data["unit"]
        
        # Calculate visualization range boundaries
        span = (high - low) if (high - low) > 0 else 10
        margin = span * 0.4
        plot_min = max(0, low - margin)
        plot_max = high + margin
        
        # Clamp values for display limits
        if val > plot_max:
            plot_max = val + margin
        if val < plot_min:
            plot_min = max(0, val - margin)
            
        # Draw Normal Range block (Green)
        fig.add_trace(go.Bar(
            y=[param],
            x=[high - low],
            base=low,
            orientation='h',
            marker=dict(
                color='rgba(46, 204, 113, 0.35)',
                line=dict(color='rgba(46, 204, 113, 0.8)', width=1)
            ),
            hoverinfo='none',
            showlegend=False
        ), row=i+1, col=1)
        
        # Mark User Value (Black Diamond / Point)
        color = 'black'
        symbol = 'diamond'
        if val < low or val > high:
            color = '#ef4444'  # Red for out-of-range
            symbol = 'x'
            
        fig.add_trace(go.Scatter(
            y=[param],
            x=[val],
            mode='markers',
            marker=dict(
                color=color,
                size=12,
                symbol=symbol,
                line=dict(color='white', width=1.5)
            ),
            hovertext=f"Your Value: {val} {unit}<br>Normal Range: {low} - {high} {unit}<br>Status: {data['status']}",
            hoverinfo='text',
            showlegend=False
        ), row=i+1, col=1)
        
        # Draw lines at low and high bounds
        fig.add_shape(
            type="line",
            x0=low, y0=-0.35, x1=low, y1=0.35,
            line=dict(color="#3b82f6", width=2, dash="dot"),
            row=i+1, col=1
        )
        fig.add_shape(
            type="line",
            x0=high, y0=-0.35, x1=high, y1=0.35,
            line=dict(color="#ef4444", width=2, dash="dot"),
            row=i+1, col=1
        )
        
        # Subplot axis configuration
        fig.update_xaxes(range=[plot_min, plot_max], row=i+1, col=1)
        fig.update_yaxes(showticklabels=False, row=i+1, col=1)
        
    fig.update_layout(
        height=max(110 * len(params), 350),
        title_text="📊 Clinical Lab Value Reference Chart",
        showlegend=False,
        template="plotly_white",
        margin=dict(l=30, r=30, t=60, b=30)
    )
    return fig
```

```python
def extract_text_from_uploaded_file(api_key, file_obj):
    """Extracts plain text from file upload, fallback to Gemini OCR if image/scan and key is provided."""
    if file_obj is None:
        return ""
        
    file_path = file_obj.name
    _, ext = os.path.splitext(file_path)
    ext = ext.lower()
    
    if ext == '.pdf':
        text = extract_text_from_pdf(file_path)
        if len(text.strip()) < 100:
            # Scanned PDF
            if not api_key:
                return "### ⚠️ Scanned PDF Detected\nThis PDF seems to be an image scan. Offline processing is not supported for scanned images.\n\n**Solution:** Please enter a Google Gemini API Key in the left panel to run cloud-based OCR." 
            else:
                return ocr_with_gemini(api_key, file_path)
        else:
            return text
    elif ext in ['.png', '.jpg', '.jpeg']:
        if not api_key:
            return "### ⚠️ Scanned Image Detected\nOffline processing is not supported for raw images.\n\n**Solution:** Please enter a Google Gemini API Key in the left panel to run cloud-based OCR."
        else:
            return ocr_with_gemini(api_key, file_path)
    else:
        return f"### ⚠️ Unsupported file type: {ext}"

def process_report_text(api_key, report_text, name_in, age_in, gender_in, date_in):
    """Runs analysis on report text, updates interactive patient info, displays results."""
    if not report_text.strip() or report_text.startswith("###"):
        # If empty or a warning message
        return (
            report_text if report_text else "### Upload a report or paste text to begin.",
            "⚠️ No active report.",
            pd.DataFrame(columns=["Lab Test", "Value", "Normal Range", "Unit", "Status"]),
            None,
            name_in or "Unknown",
            age_in or "Unknown",
            gender_in or "Unknown",
            date_in or "Unknown",
            "No report active."
        )
        
    # Extract metadata from text if input values are default or empty
    meta_extracted = extract_metadata(report_text)
    
    name = name_in if name_in and name_in != "Unknown" else meta_extracted.get("Name", "Unknown")
    age = age_in if age_in and age_in != "Unknown" else meta_extracted.get("Age", "Unknown")
    gender = gender_in if gender_in and gender_in != "Unknown" else meta_extracted.get("Gender", "Unknown")
    date = date_in if date_in and date_in != "Unknown" else meta_extracted.get("Date", "Unknown")
    
    metadata = {"Name": name, "Age": age, "Gender": gender, "Date": date}
    
    # Parse lab tests
    parsed_results = parse_lab_values_from_text(report_text)
    
    # Summary generation
    if api_key:
        summary_text = analyze_report_with_gemini(api_key, report_text, str(parsed_results))
    else:
        summary_text = generate_offline_summary(parsed_results, metadata)
        
    # Table mapping
    table_data = []
    for test_name, data in parsed_results.items():
        table_data.append({
            "Lab Test": test_name,
            "Value": f"{data['value']}",
            "Normal Range": f"{data['low']} - {data['high']}",
            "Unit": data['unit'],
            "Status": data['status']
        })
        
    df = pd.DataFrame(table_data)
    if df.empty:
        df = pd.DataFrame(columns=["Lab Test", "Value", "Normal Range", "Unit", "Status"])
        
    # Warnings and Subplots
    red_flags_md = generate_red_flags_markdown(parsed_results)
    plotly_fig = generate_plotly_charts(parsed_results)
    
    # Chat Context
    chat_context = f"Patient Name: {name}, Age: {age}, Gender: {gender}, Date: {date}.\n"
    if parsed_results:
        chat_context += "Extracted lab parameters:\n"
        for test_name, data in parsed_results.items():
            chat_context += f"- {test_name}: {data['value']} {data['unit']} (Ref: {data['low']}-{data['high']}) Status: {data['status']}\n"
    else:
        chat_context += "No lab values could be parsed cleanly by the offline rule engine."
        
    return (
        summary_text,
        red_flags_md,
        df,
        plotly_fig,
        name,
        age,
        gender,
        date,
        chat_context
    )

def load_sample_report():
    """Loads a standard mock report text to show how it functions offline instantly."""
    return """PATIENT LAB REPORT
-------------------------
Patient Name: John Doe
Age: 42
Gender: Male
Date of Report: 2026-07-01

TEST RESULTS:
Glucose (Fasting)      108.5   mg/dL     (70.0 - 99.0)
Hemoglobin             11.4    g/dL      (12.0 - 17.5)
White Blood Cells (WBC) 12.8    x10^3/uL  (4.5 - 11.0)
Total Cholesterol      245.0   mg/dL     (< 200.0)
Creatinine             0.9     mg/dL     (0.6 - 1.3)
TSH (Thyroid)          4.8     mIU/L     (0.4 - 4.0)
Potassium              5.5     mEq/L     (3.5 - 5.2)
"""

# HTML Header
HEADER_HTML = """
<div style="text-align: center; margin-bottom: 25px; padding: 15px; border-radius: 12px; background: rgba(15, 23, 42, 0.4); border: 1px solid rgba(6, 182, 212, 0.15);">
    <h1 style="margin: 0; font-size: 2.8rem; font-weight: 900; background: linear-gradient(135deg, #22d3ee, #34d399); -webkit-background-clip: text; -webkit-text-fill-color: transparent;">🩺 Medical Report AI Agent</h1>
    <p style="font-size: 1.15rem; color: #94a3b8; margin-top: 8px; font-weight: 500;">A state-of-the-art clinical dashboard that extracts lab values, flags abnormalities, renders reference ranges, and explains reports using Google Gemini.</p>
</div>
"""

CSS_STYLING = """
/* Modern Premium Dark Glassmorphism Styling */
.gradio-container {
    background-color: #0b0f19 !important;
    background-image: 
        radial-gradient(circle at 10% 20%, rgba(6, 182, 212, 0.04) 0%, transparent 40%),
        radial-gradient(circle at 90% 80%, rgba(16, 185, 129, 0.04) 0%, transparent 40%) !important;
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, "Helvetica Neue", Arial, sans-serif !important;
}

/* Glassmorphism panel cards */
.gr-group, .gr-box, .gr-form {
    background-color: rgba(17, 25, 40, 0.7) !important;
    border: 1px solid rgba(56, 189, 248, 0.12) !important;
    border-radius: 16px !important;
    box-shadow: 0 8px 32px 0 rgba(0, 0, 0, 0.45) !important;
    backdrop-filter: blur(12px) !important;
    -webkit-backdrop-filter: blur(12px) !important;
    margin-bottom: 20px !important;
}

/* Text areas and inputs style */
input, textarea, select {
    background-color: rgba(15, 23, 42, 0.6) !important;
    border: 1px solid rgba(56, 189, 248, 0.2) !important;
    color: #f8fafc !important;
    border-radius: 10px !important;
    padding: 10px !important;
}
input:focus, textarea:focus {
    border-color: #22d3ee !important;
    box-shadow: 0 0 0 2px rgba(34, 211, 238, 0.25) !important;
}

/* Styling for interactive patient metadata textboxes specifically */
.interactive-info-textbox input {
    background-color: rgba(6, 182, 212, 0.05) !important;
    border: 1px solid rgba(6, 182, 212, 0.25) !important;
    font-weight: 600 !important;
    color: #22d3ee !important;
}

/* Glassy warning panel */
.red-flag-box {
    background-color: rgba(239, 68, 68, 0.08) !important;
    border-left: 5px solid #ef4444 !important;
    border-radius: 8px;
    padding: 15px;
    margin: 10px 0;
    border: 1px solid rgba(239, 68, 68, 0.15);
}

/* Primary neon-teal buttons */
button.primary {
    background: linear-gradient(135deg, #06b6d4 0%, #0d9488 100%) !important;
    color: #ffffff !important;
    border: none !important;
    border-radius: 10px !important;
    font-weight: 700 !important;
    letter-spacing: 0.5px !important;
    box-shadow: 0 4px 15px rgba(6, 182, 212, 0.35) !important;
    transition: all 0.3s cubic-bezier(0.4, 0, 0.2, 1) !important;
}
button.primary:hover {
    transform: translateY(-2px) !important;
    box-shadow: 0 8px 25px rgba(6, 182, 212, 0.55) !important;
    background: linear-gradient(135deg, #0891b2 0%, #0f766e 100%) !important;
}

/* Navigation buttons / Tabs styling */
.tab-nav {
    border-bottom: 2px solid rgba(255, 255, 255, 0.06) !important;
    margin-bottom: 20px !important;
}
.tab-nav button {
    font-size: 1.05rem !important;
    font-weight: 700 !important;
    color: #64748b !important;
    background: transparent !important;
    border: none !important;
    transition: all 0.25s ease !important;
}
.tab-nav button.selected {
    color: #22d3ee !important;
    border-bottom: 3px solid #22d3ee !important;
    background: rgba(34, 211, 238, 0.04) !important;
}

/* Searchable table style */
.gr-dataframe {
    border-radius: 12px !important;
    border: 1px solid rgba(56, 189, 248, 0.1) !important;
    overflow: hidden !important;
}
.gr-dataframe th {
    background-color: rgba(6, 182, 212, 0.1) !important;
    color: #22d3ee !important;
    font-weight: 700 !important;
    border-bottom: 1px solid rgba(56, 189, 248, 0.1) !important;
}
"""

with gr.Blocks() as demo:
    gr.HTML(HEADER_HTML)
    
    # Internal states
    chat_context_state = gr.State("")
    
    with gr.Row():
        # Setup & Control Column (Left)
        with gr.Column(scale=1):
            gr.Markdown("### ⚙️ Configuration & Inputs")
            api_key_input = gr.Textbox(
                label="Google Gemini API Key (Optional)", 
                placeholder="Paste Gemini API Key here...", 
                type="password"
            )
            gr.Markdown("*Offline Mode parses digital PDFs locally without a key! Load the Sample Report to see it run.*")
            
            # Input selection modes
            with gr.Tabs():
                with gr.TabItem("📁 Document Upload"):
                    file_uploader = gr.File(
                        label="Upload Lab Report (PDF, PNG, JPG)", 
                        file_types=[".pdf", ".png", ".jpg", ".jpeg"]
                    )
                    
                with gr.TabItem("✍️ Direct Text Editor"):
                    gr.Markdown("Paste report content directly or edit extracted text below:")
                    
            # Common Textbox representing the report text content
            report_text_input = gr.Textbox(
                label="Active Report Text Content",
                placeholder="Extracted report text or pasted text will appear here. You can manually edit this text!",
                lines=10,
                max_lines=20,
                interactive=True
            )
            
            with gr.Row():
                load_sample_btn = gr.Button("💡 Load Sample", variant="secondary")
                analyze_btn = gr.Button("🔍 Analyze Report", variant="primary")
                
            gr.Markdown(
                """
                > [!WARNING]
                > **Medical Disclaimer:** This tool is for informational purposes only. It is not diagnostic. Consult a licensed physician for healthcare advice.
                """
            )
            
        # Result Analysis Column (Right)
        with gr.Column(scale=2.2):
            with gr.Tabs():
                with gr.TabItem("📊 Analysis Dashboard"):
                    with gr.Row():
                        with gr.Column(scale=2):
                            # Patient Metadata Card (Fully Interactive now!)
                            with gr.Group():
                                gr.Markdown("### 👤 Patient Information (Editable)")
                                with gr.Row():
                                    patient_name_val = gr.Textbox(label="Patient Name", value="Unknown", interactive=True, elem_classes="interactive-info-textbox")
                                    patient_age_val = gr.Textbox(label="Age", value="Unknown", interactive=True, elem_classes="interactive-info-textbox")
                                    patient_gender_val = gr.Textbox(label="Gender", value="Unknown", interactive=True, elem_classes="interactive-info-textbox")
                                    report_date_val = gr.Textbox(label="Report Date", value="Unknown", interactive=True, elem_classes="interactive-info-textbox")
                                    
                        with gr.Column(scale=1):
                            # Red flags warning quick-view
                            red_flags_quick = gr.Markdown("### Abnormality Check\n*Provide report text or file to start.*")
                            
                    gr.Markdown("---")
                    
                    # AI Report Summary
                    gr.Markdown("### 📋 Executive Report Summary")
                    report_summary = gr.Markdown("*Your clinical report summary and explanations will generate here.*")
                    
                    gr.Markdown("---")
                    
                    # Extracted Lab values Table
                    gr.Markdown("### 🧬 Extracted Clinical Markers")
                    lab_table = gr.DataFrame(
                        headers=["Lab Test", "Value", "Normal Range", "Unit", "Status"],
                        datatype=["str", "str", "str", "str", "str"],
                        interactive=False
                    )
                    
                    gr.Markdown("---")
                    
                    # Reference chart (Fixed to gr.Plot())
                    gr.Markdown("### 📈 Visual Reference Intervals")
                    plotly_chart = gr.Plot()
                    
                with gr.TabItem("💬 Ask the Medical Agent"):
                    gr.Markdown("### 🤖 Medical Chat Assistant")
                    gr.Markdown("Ask questions regarding your report, dietary tips, or clinical terms. *Note: Chatbot requires Gemini API Key.*")
                    
                    chatbot = gr.Chatbot(height=450)
                    chat_input = gr.Textbox(placeholder="Ask me anything about your report findings...", label="Your Question")
                    
                    with gr.Row():
                        submit_chat_btn = gr.Button("Send", variant="primary", scale=4)
                        clear_chat_btn = gr.Button("Clear Chat", variant="stop", scale=1)
                        
                with gr.TabItem("📖 Clinical Dictionary & Guide"):
                    gr.Markdown("### 📘 Lab Parameter Reference Guide")
                    gr.Markdown("Understand what common lab tests represent and their normal ranges:")
                    
                    for name, details in LAB_DICTIONARY.items():
                        with gr.Accordion(f"{name} ({details['unit']})", open=False):
                            gr.Markdown(
                                f"""
                                **Standard Reference Range:** `{details['low']} - {details['high']} {details['unit']}`
                                
                                **Clinical Meaning:**
                                {details['description']}
                                """
                            )
                            
    # EVENT HOOKS
    
    # 1. When file is uploaded: extract text, then run process
    file_uploader.upload(
        fn=extract_text_from_uploaded_file,
        inputs=[api_key_input, file_uploader],
        outputs=[report_text_input]
    ).then(
        fn=process_report_text,
        inputs=[api_key_input, report_text_input, patient_name_val, patient_age_val, patient_gender_val, report_date_val],
        outputs=[report_summary, red_flags_quick, lab_table, plotly_chart, patient_name_val, patient_age_val, patient_gender_val, report_date_val, chat_context_state]
    )
    
    # 2. When Load Sample is clicked: fill sample report text, then run process
    load_sample_btn.click(
        fn=load_sample_report,
        inputs=[],
        outputs=[report_text_input]
    ).then(
        fn=process_report_text,
        inputs=[api_key_input, report_text_input, patient_name_val, patient_age_val, patient_gender_val, report_date_val],
        outputs=[report_summary, red_flags_quick, lab_table, plotly_chart, patient_name_val, patient_age_val, patient_gender_val, report_date_val, chat_context_state]
    )
    
    # 3. When Analyze button is clicked: process current text box content
    analyze_btn.click(
        fn=process_report_text,
        inputs=[api_key_input, report_text_input, patient_name_val, patient_age_val, patient_gender_val, report_date_val],
        outputs=[report_summary, red_flags_quick, lab_table, plotly_chart, patient_name_val, patient_age_val, patient_gender_val, report_date_val, chat_context_state]
    )
    
    # 4. Chatbot interaction
    submit_chat_btn.click(
        fn=chatbot_respond,
        inputs=[api_key_input, chat_input, chatbot, chat_context_state],
        outputs=[chat_input, chatbot]
    )
    chat_input.submit(
        fn=chatbot_respond,
        inputs=[api_key_input, chat_input, chatbot, chat_context_state],
        outputs=[chat_input, chatbot]
    )
    clear_chat_btn.click(lambda: None, None, chatbot, queue=False)
```
